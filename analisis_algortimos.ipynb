```python
import time
import random
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para los gráficos
sns.set_theme(style="whitegrid")
```

```python
# Definición de los algoritmos de ordenamiento

def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n-i-1):
            if lista[j] > lista[j+1]:
                lista[j], lista[j+1] = lista[j+1], lista[j]
    return lista

def merge_sort(lista):
    if len(lista) <= 1:
        return lista
    medio = len(lista) // 2
    izquierda = merge_sort(lista[:medio])
    derecha = merge_sort(lista[medio:])
    
    # Combinar sublistas ordenadas
    resultado = []
    i = j = 0
    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] < derecha[j]:
            resultado.append(izquierda[i])
            i += 1
        else:
            resultado.append(derecha[j])
            j += 1
    resultado.extend(izquierda[i:])
    resultado.extend(derecha[j:])
    return resultado

print("¡Funciones de ordenamiento cargadas con éxito!")
```

```python
# Lista de tamaños de datos que vamos a evaluar (N)
tamanos = [100, 500, 1000, 2000]
resultados = []

print("Ejecutando mediciones...")

for n in tamanos:
    # Generar una lista de números aleatorios de tamaño N
    lista_original = [random.randint(0, 10000) for _ in range(n)]
    
    # 1. Medir Bubble Sort O(n^2)
    inicio = time.time()
    bubble_sort(lista_original.copy())
    fin = time.time()
    resultados.append({
        "Tamaño (N)": n, 
        "Tiempo (s)": fin - inicio, 
        "Algoritmo": "Bubble Sort - O(n²)"
    })
    
    # 2. Medir Merge Sort O(n log n)
    inicio = time.time()
    merge_sort(lista_original.copy())
    fin = time.time()
    resultados.append({
        "Tamaño (N)": n, 
        "Tiempo (s)": fin - inicio, 
        "Algoritmo": "Merge Sort - O(n log n)"
    })

# Convertir la lista de diccionarios en un DataFrame de Pandas
df_resultados = pd.DataFrame(resultados)

# Mostrar la tabla formateada
print("¡Mediciones completadas!")
df_resultados
```

```python
plt.figure(figsize=(10, 6))

# Gráfico de líneas con puntos marcadores
sns.lineplot(
    data=df_resultados, 
    x="Tamaño (N)", 
    y="Tiempo (s)", 
    hue="Algoritmo", 
    style="Algoritmo",
    markers=True, 
    dashes=False,
    linewidth=2.5
)

# Títulos y etiquetas técnicas
plt.title("Comparación de Complejidad Temporal (Big-O)", fontsize=14, fontweight='bold')
plt.xlabel("Cantidad de Datos (N)", fontsize=12)
plt.ylabel("Tiempo de Ejecución (segundos)", fontsize=12)

# Mostrar el gráfico
plt.show()
```

## Conclusiones del Experimento

1. **Eficiencia en Escala:** Para arreglos pequeños ($N \le 500$), la diferencia de tiempo entre ambos algoritmos es prácticamente imperceptible.
2. **Crecimiento Cuadrático vs Logarítmico:** A medida que $N$ crece hacia $2000$ elementos, `Bubble Sort` incrementa su tiempo de forma cuadrática, mientras que `Merge Sort` mantiene un desempeño óptimo gracias a la estrategia de *Divide y Vencerás*.
3. **Recomendación:** En escenarios de análisis de grandes volúmenes de datos, el uso de algoritmos $O(N^2)$ debe ser descartado en favor de enfoques logarítmicos o vectorizados.
